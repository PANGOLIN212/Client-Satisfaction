```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

F = 1.0        # size factor: 1.0 = normal, 0.65 = 35% smaller (matches your Figure 1)

df = pd.DataFrame({
    "Facility": ["A"]*3 + ["B"]*3 + ["C"]*3,
    "Month":    ["Jan", "Feb", "Mar"] * 3,
    "Wait":     [48, 52, 55,  35, 33, 31,  42, 39, 36],
    "Sat":      [3.6, 3.4, 3.2,  4.1, 4.2, 4.3,  3.8, 4.0, 4.2],
})
```

```python
COL   = {"A": "#D55E00", "B": "#0072B2", "C": "#009E73"}   # colour-blind-safe palette
MRK   = {"A": "o", "B": "s", "C": "^"}                      # circle, square, triangle
STAFF = {"A": "Low", "B": "Medium", "C": "High"}
```

```python
plt.rcParams["font.size"] = 10 * F
fig, ax = plt.subplots(figsize=(6.6 * F, 3.9 * F), dpi=200)
for f in "ABC":
    d = df[df.Facility == f]
    ax.scatter(d.Wait, d.Sat, s=80 * F**2, marker=MRK[f], color=COL[f], zorder=3,
               label=f"Facility {f} ({STAFF[f]} staffing)")
    for _, row in d.iterrows():
        dx, dy = (-4, -15) if f == "B" else (6, 6)
        ax.annotate(row.Month, (row.Wait, row.Sat), textcoords="offset points",
                    xytext=(dx * F, dy * F), fontsize=8 * F, color=COL[f],
                    ha="center" if f == "B" else "left")
        slope, intercept = np.polyfit(df.Wait, df.Sat, 1)
r = np.corrcoef(df.Wait, df.Sat)[0, 1]
xs = np.array([29, 57])
ax.plot(xs, slope * xs + intercept, "--", color="#777777", lw=1 * F, zorder=1,
        label=f"Overall linear trend (r = {r:.2f}, n = 9)".replace("-", "\u2212"))

ax.set_xlim(28, 58)
ax.set_ylim(3.0, 4.5)
ax.set_xlabel("Average waiting time (minutes)")
ax.set_ylabel("Client satisfaction (mean rating, 1\u20135 scale)")
ax.set_title("Longer average waits coincided with lower client satisfaction",
             fontsize=7.5 * F, fontweight="bold", loc="left")
ax.grid(color="#E3E3E3", lw=0.6 * F)
ax.set_axisbelow(True)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False, fontsize=8 * F, loc="upper right")
fig.tight_layout()
fig.savefig("fig2.png")
```
